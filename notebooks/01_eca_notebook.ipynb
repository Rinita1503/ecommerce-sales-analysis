```python
#1. Importing libraries:
import pandas as pd
import numpy as np

# Display_settings: For better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)
```

```python
#2. Load dataset:
df = pd.read_csv('sample_superstore.csv', encoding='latin1')
df.to_csv('fixed_file.csv', index=False, encoding='utf-8')
```

```python
#3. Preview data:
print("\n--- First 5 rows ---")
display(df.head())

print("\n--- Last 5 rows ---")
display(df.tail())
```

```python
#4. Dataset structure:
print("\n--- Dataset Info ---")
df.info()
```

```python
# 5. Statistical summary
print("\n--- Statistical Summary ---")
display(df.describe())
```

```python
# 6. Check missing values
print("\n--- Missing Values ---")
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage
}).sort_values(by="Missing Values", ascending=False)

display(missing_df)
```

```python
# 7. Check duplicates
print("\n--- Duplicate Rows ---")
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
```

```python
# 8. Convert date columns
print("\n--- Converting Date Columns ---")
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

print("Date conversion completed.")
```

```python
# 9. Basic business checks
print("\n--- Business Data Checks ---")

# Negative profit (loss-making sales)
negative_profit = df[df["Profit"] < 0].shape[0]
print(f"Transactions with negative profit: {negative_profit}")

# Discount impact
high_discount = df[df["Discount"] > 0.5].shape[0]
print(f"Transactions with high discount (>50%): {high_discount}")

# Sales overview
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print(f"\nTotal Sales: {total_sales:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
```

```python
# 10. Unique counts
print("\n--- Unique Counts ---")
print(f"Unique Customers: {df['Customer ID'].nunique()}")
print(f"Unique Products: {df['Product ID'].nunique()}")
print(f"Unique Categories: {df['Category'].nunique()}")
print(f"Unique Regions: {df['Region'].nunique()}")
```

```python
# 11. Top categories
print("\n--- Sales by Category ---")
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
display(category_sales)
```
