```python
# 1. Import libraries:
import pandas as pd
import numpy as np
```

```python
#2. Load dataset:
df = pd.read_csv('sample_superstore.csv', encoding='latin1')
df.to_csv('fixed_file.csv', index=False, encoding='utf-8')
```

```python
# 3. Initial shape:
print("\nInitial Shape:", df.shape)
```

# Data Cleaning

```python
# 4. Remove duplicates:
duplicates = df.duplicated().sum()
print(f"/nDuplicate rows before removal: {duplicates}")
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
#5. Convert date columns:
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')
```

```python
# 6. Handle missing values:
print("\nMissing values before handling:")
print(df.isnull().sum())
```

```python
# 7. For this dataset, usually minimal missing values:
df = df.dropna()
print("\nShape after dropping missing values:", df.shape)
```

# Feature Engineering

```python
# 8. Create Profit Margin:
df['Profit Margin'] = df['Profit'] / df['Sales']
print(df)
```

```python
# 9. Create Order Processing Time:
df['Shipping Time (Days)'] = (df['Ship Date'] - df['Order Date']).dt.days
print(df)
```

```python
# 10. Extract Date Features:
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Month Name'] = df['Order Date'].dt.month_name()
print(df)
```

```python
#11. Create Sales Segments:
df['Sales Segment'] = pd.qcut(df['Sales'], q=3, labels=['Low', 'Medium', 'High'])
print(df)
```

```python
# 12. Discount Flag
df['High Discount'] = df['Discount'].apply(lambda x: 1 if x > 0.2 else 0)
print(df)
```
